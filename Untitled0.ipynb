```python
# peak spending month vs low spending month
peak=[4,5,7,9,10]
df["Peak_month"]=df["Month"].apply(lambda x:"peak_month" if x in peak else "normal_month")
```

```python
# check the statistical analysis
#H0: there is no difference between the spend and the workation trips
#HA: people spend in peak months is higher than the rest
# two sample independant t- test
peak=df[df["Peak_month"]=="peak_month"]
normal=df[df["Peak_month"]=="normal_month"]
import scipy.stats as stats
stats.ttest_ind(peak["Per Person Price"],normal["Per Person Price"])
```

```python
df["Destination"].unique()
#total no of places travelled
# creating new feature called no of destinations travelled
def removing(x):
  if " and " or " " in x:
    # do replace and with | and split on |
    x.replace("and","|")
    x.split("|")
    return len(x)
  else:
    return len(x.split("|"))

```

```python
df["no_of_destinations"]=df["Destination"].apply(removing)
```

```python
# repeat the process for airline and places covered
df["Airline_places"]=df["Airline"].apply(removing)
```

```python
# extracting the numerics using regex and summing them up

import re

a = '1N Colombo . 2N Dambulla . 2N Kandy . 1N Nuwara Eliya . 1N Yala . 2N Bentota'
numbers =sum(pd.to_numeric( re.findall(r'[0-9]', a)))
print(numbers)
```

```python
#regex function applied for itenarary
def duration(x):
  return np.sum(pd.to_numeric(re.findall(r"\d",x)))
df["duration"]=df["Itinerary"].apply(duration)
```

```python
df["Hotel Details"].unique()# not available in case of ratings we can mark it as 0 and impute with average later
```

```python
def convert(text):
  text = text.replace(':One',':1')
  text = text.replace(':Two',':2')
  text = text.replace(':Three',':3')
  text = text.replace(':Four',':4')
  text = text.replace(':Five',':5')
  return text
```

```python
df["Hotel Details_new"]=df["Hotel Details"].apply(convert)
```

```python
df["Hotel Details_new"].value_counts()
```

```python
def ratings(text):
  return (np.mean(pd.to_numeric(re.findall(r"\d+\.?\d*",text))))
```

```python
df["Hotel_ratings"]=df["Hotel Details_new"].apply(ratings)
```
